## Bài 1. Các số chia hết cho 7 nhưng không chia hết cho 5

```python
def bai_01():
    ket_qua = [str(n) for n in range(2000, 3201) if n % 7 == 0 and n % 5 != 0]
    return ", ".join(ket_qua)

print(bai_01())
```

## Bài 2. Tính giai thừa của một số

```python
def bai_02(n):
    if n < 0:
        return "Không tính được giai thừa của số âm."
    ket_qua = 1
    for i in range(2, n + 1):
        ket_qua *= i
    return ket_qua

n = int(input("Nhập số nguyên n: "))
print(bai_02(n))
```

## Bài 3. Tạo dictionary từ 1 đến n với giá trị bình phương

```python
def bai_03(n):
    return {i: i * i for i in range(1, n + 1)}

n = int(input("Nhập số nguyên n: "))
print(bai_03(n))
```

## Bài 4. Chuyển chuỗi phân tách bằng dấu phẩy thành list và tuple

```python
def bai_04(chuoi):
    danh_sach = [phan_tu.strip() for phan_tu in chuoi.split(",")]
    bo_gia_tri = tuple(danh_sach)
    return f"{danh_sach}\n{bo_gia_tri}"

chuoi = input("Nhập các số cách nhau bằng dấu phẩy: ")
print(bai_04(chuoi))
```

## Bài 5. Class với getString và printString

```python
class ChuoiVanBan:
    def __init__(self):
        self.chuoi = ""

    def getString(self, chuoi):
        self.chuoi = chuoi

    def printString(self):
        return self.chuoi.upper()


def bai_05(chuoi):
    doi_tuong = ChuoiVanBan()
    doi_tuong.getString(chuoi)
    return doi_tuong.printString()

chuoi = input("Nhập chuỗi: ")
print(bai_05(chuoi))
```

## Bài 6. Method tính bình phương của một số

```python
class MayTinhBinhPhuong:
    def binh_phuong(self, n):
        return n ** 2


def bai_06(n):
    may_tinh = MayTinhBinhPhuong()
    return may_tinh.binh_phuong(n)

n = float(input("Nhập số cần tính bình phương: "))
print(bai_06(n))
```

## Bài 7. Xem tài liệu hàm bằng __doc__

```python
def ham_tu_dinh_nghia(x):
    """Trả về bình phương của x."""
    return x ** 2


def bai_07():
    ham = [abs, int, input, ham_tu_dinh_nghia]
    return "\n\n".join(f"{f.__name__}:\n{f.__doc__}" for f in ham)

print(bai_07())
```

## Bài 8. Tham số lớp và tham số instance cùng tên

## Bài 9. Tính tuổi dựa trên ngày sinh

```python
from datetime import datetime, date


def bai_09(ngay_sinh):
    sinh = datetime.strptime(ngay_sinh, "%Y/%m/%d").date()
    hom_nay = date.today()
    if sinh > hom_nay:
        return "Ngày sinh không được ở tương lai."
    tuoi = hom_nay.year - sinh.year
    if (hom_nay.month, hom_nay.day) < (sinh.month, sinh.day):
        tuoi -= 1
    return f"Tuổi hiện tại: {tuoi}"

ngay_sinh = input("Nhập ngày sinh (yyyy/mm/dd): ")
print(bai_09(ngay_sinh))
```

## Bài 10. Tính tiền lương có giờ làm thêm

```python
def bai_10(so_gio, luong_gio):
    if so_gio < 0 or luong_gio < 0:
        return "Số giờ và mức lương không được âm."
    gio_thuong = min(so_gio, 44)
    gio_them = max(so_gio - 44, 0)
    tong_luong = gio_thuong * luong_gio + gio_them * luong_gio * 1.5
    return f"Tiền lương thực lĩnh: {tong_luong:.2f}"

so_gio = float(input("Nhập số giờ làm trong tuần: "))
luong_gio = float(input("Nhập thù lao mỗi giờ tiêu chuẩn: "))
print(bai_10(so_gio, luong_gio))
```

## Bài 11. Tính Q theo công thức với nhiều giá trị D

```python
from math import sqrt


def bai_11(chuoi_d):
    C, H = 50, 30
    cac_gia_tri = [float(d.strip()) for d in chuoi_d.split(",") if d.strip()]
    if any(d < 0 for d in cac_gia_tri):
        return "Các giá trị D phải không âm."
    ket_qua = [str(round(sqrt(2 * C * d / H))) for d in cac_gia_tri]
    return ", ".join(ket_qua)

chuoi_d = input("Nhập các giá trị D, cách nhau bằng dấu phẩy: ")
print(bai_11(chuoi_d))
```

## Bài 12. Tạo mảng hai chiều với phần tử i × j

```python
def bai_12(x, y):
    if x < 0 or y < 0:
        return "Kích thước mảng phải không âm."
    return [[i * j for j in range(y)] for i in range(x)]

x = int(input("Nhập số hàng X: "))
y = int(input("Nhập số cột Y: "))
print(bai_12(x, y))
```

## Bài 13. Sắp xếp các từ theo thứ tự bảng chữ cái

```python
def bai_13(chuoi):
    cac_tu = [tu.strip() for tu in chuoi.split(",") if tu.strip()]
    return ", ".join(sorted(cac_tu))

chuoi = input("Nhập các từ, cách nhau bằng dấu phẩy: ")
print(bai_13(chuoi))
```

## Bài 14. Đổi nhiều dòng văn bản thành chữ hoa

```python
def bai_14(cac_dong):
    return "\n".join(dong.upper() for dong in cac_dong)

cac_dong = []
print("Nhập từng dòng (Enter trên dòng trống để kết thúc):")
while True:
    dong = input()
    if dong == "":
        break
    cac_dong.append(dong)
print(bai_14(cac_dong))
```

## Bài 15. Xóa từ trùng lặp và sắp xếp

## Bài 16. Lọc số nhị phân bốn chữ số chia hết cho 5

```python
def bai_16(chuoi):
    cac_so = [so.strip() for so in chuoi.split(",") if so.strip()]
    for so in cac_so:
        if len(so) != 4 or any(ch not in "01" for ch in so):
            return f"'{so}' không phải số nhị phân 4 chữ số."
    return ",".join(so for so in cac_so if int(so, 2) % 5 == 0)

chuoi = input("Nhập các số nhị phân 4 chữ số, cách nhau bằng dấu phẩy: ")
print(bai_16(chuoi))
```

## Bài 17. Các số từ 1000 đến 3000 chỉ chứa chữ số chẵn

```python
def bai_17():
    ket_qua = [str(n) for n in range(1000, 3001)
               if all(int(chu_so) % 2 == 0 for chu_so in str(n))]
    return ",".join(ket_qua)

print(bai_17())
```

## Bài 18. Đếm chữ cái và chữ số trong câu

```python
def bai_18(chuoi):
    so_chu_cai = sum(ky_tu.isalpha() for ky_tu in chuoi)
    so_chu_so = sum(ky_tu.isdigit() for ky_tu in chuoi)
    return f"Số chữ cái là: {so_chu_cai}\nSố chữ số là: {so_chu_so}"

chuoi = input("Nhập một câu: ")
print(bai_18(chuoi))
```

## Bài 19. Đếm chữ hoa và chữ thường

```python
def bai_19(chuoi):
    so_chu_hoa = sum(ky_tu.isupper() for ky_tu in chuoi)
    so_chu_thuong = sum(ky_tu.islower() for ky_tu in chuoi)
    return f"Chữ hoa: {so_chu_hoa}\nChữ thường: {so_chu_thuong}"

chuoi = input("Nhập một câu: ")
print(bai_19(chuoi))
```

## Bài 20. Tính tổng a + aa + aaa + aaaa

```python
def bai_20(a):
    if len(a) != 1 or not a.isdigit():
        return "a phải là một chữ số từ 0 đến 9."
    return sum(int(a * i) for i in range(1, 5))

a = input("Nhập chữ số a: ").strip()
print(bai_20(a))
```

## Bài 21. Lọc các số lẻ trong danh sách

```python
def bai_21(chuoi):
    cac_so = [int(so.strip()) for so in chuoi.split(",") if so.strip()]
    return ",".join(str(so) for so in cac_so if so % 2 != 0)

chuoi = input("Nhập các số nguyên, cách nhau bằng dấu phẩy: ")
print(bai_21(chuoi))
```

## Bài 22. Tính số dư tài khoản từ các giao dịch

```python
def bai_22(cac_giao_dich):
    so_du = 0
    for giao_dich in cac_giao_dich:
        loai, so_tien = giao_dich.split()
        so_tien = int(so_tien)
        if so_tien < 0:
            return "Số tiền giao dịch phải không âm."
        if loai.upper() == "D":
            so_du += so_tien
        elif loai.upper() == "W":
            so_du -= so_tien
        else:
            return f"Loại giao dịch không hợp lệ: {loai}"
    return so_du

cac_giao_dich = []
print("Nhập các giao dịch dạng D 100 hoặc W 200 (Enter trống để kết thúc):")
while True:
    giao_dich = input()
    if not giao_dich.strip():
        break
    cac_giao_dich.append(giao_dich)
print(bai_22(cac_giao_dich))
```

## Bài 23. Kiểm tra và biến đổi dãy thành đối xứng

```python
def kiem_tra_doi_xung(day_so):
    return day_so == day_so[::-1]


def sx_doi_xung(day_so):
    ket_qua = day_so.copy()
    cac_cap_khac_nhau = [i for i in range(len(ket_qua) // 2)
                         if ket_qua[i] != ket_qua[-1 - i]]
    if len(cac_cap_khac_nhau) > 1:
        return None
    if cac_cap_khac_nhau:
        i = cac_cap_khac_nhau[0]
        ket_qua[-1 - i] = ket_qua[i]
    return ket_qua


def bai_23(day_so):
    if kiem_tra_doi_xung(day_so):
        return f"Dãy đã đối xứng: {day_so}"
    ket_qua = sx_doi_xung(day_so)
    if ket_qua is None:
        return "Không thể đối xứng bằng cách thay đổi tối đa một phần tử."
    return f"Dãy sau khi điều chỉnh: {ket_qua}"

day_so = list(map(int, input("Nhập dãy số nguyên, cách nhau bằng dấu cách: ").split()))
print(bai_23(day_so))
```
