## Bài 40. Tổng các số chia hết cho 3 hoặc 5 trong [M, N]

```python
def bai_40(M, N):
    if not 0 < M < N:
        return "Cần nhập hai số nguyên dương M < N."
    tong = 0
    for i in range(M, N + 1):
        if i % 3 == 0 or i % 5 == 0:
            tong += i
    return f"Kết quả: {tong}"

M = int(input("Nhập M: "))
N = int(input("Nhập N: "))
print(bai_40(M, N))
```

## Bài 41. Tổng các số chia hết cho 2 và 3 trong [M, N]

## Bài 42. Liệt kê các ước của n

```python
def bai_42(n):
    if n < 1:
        return "n phải là số nguyên dương."
    cac_uoc = []
    for i in range(1, n + 1):
        if n % i == 0:
            cac_uoc.append(i)
    return "\n".join(map(str, cac_uoc))

n = int(input("Nhập số n: "))
print(bai_42(n))
```

## Bài 43. Đếm số ước của n

```python
def bai_43(n):
    import math
    if n < 1:
        return "n phải là số nguyên dương."
    dem = 0
    for i in range(1, math.isqrt(n) + 1):
        if n % i == 0:
            dem += 1 if i * i == n else 2
    return f"{n} co {dem} uoc"

n = int(input("Nhập n: "))
print(bai_43(n))
```

## Bài 44. Đếm số trong khoảng (10, 20)

```python
def bai_44(cac_so):
    dem = 0
    for so in cac_so:
        if 10 < so < 20:
            dem += 1
    return f"Số các số >10 và <20 là: {dem}"

N = int(input("Nhập N: "))
cac_so = []
for i in range(N):
    so = float(input(f"Nhập số thứ {i + 1}: "))
    cac_so.append(so)
print(bai_44(cac_so))
```

## Bài 45. Kiểm tra số nguyên tố

```python
def bai_45(n):
    import math
    if n < 2:
        return f"{n} không là số nguyên tố"
    for i in range(2, math.isqrt(n) + 1):
        if n % i == 0:
            return f"{n} không là số nguyên tố"
    return f"{n} là số nguyên tố"

n = int(input("Nhập số n: "))
print(bai_45(n))
```

## Bài 46. Kiểm tra số hoàn hảo

```python
def bai_46(n):
    tong_uoc = 0
    if n > 1:
        for i in range(1, n):
            if n % i == 0:
                tong_uoc += i
    return f"{n} là số hoàn hảo" if n > 1 and tong_uoc == n else f"{n} không là số hoàn hảo"

n = int(input("Nhập n: "))
print(bai_46(n))
```

## Bài 47. Các số hoàn hảo nhỏ hơn n

```python
def bai_47(n):
    import math
    cac_so_hoan_hao = []
    for so in range(2, n):
        tong_uoc = 1
        for i in range(2, math.isqrt(so) + 1):
            if so % i == 0:
                tong_uoc += i
                if i != so // i:
                    tong_uoc += so // i
        if tong_uoc == so:
            cac_so_hoan_hao.append(so)
    return "\n".join(map(str, cac_so_hoan_hao)) if cac_so_hoan_hao else "Không có số hoàn hảo nào."

n = int(input("Nhập số n: "))
print(bai_47(n))
```

## Bài 48. Số chính phương

```python
def bai_48(n):
    import math
    la_so_chinh_phuong = n > 0 and math.isqrt(n) ** 2 == n
    return f"{n} là số chính phương" if la_so_chinh_phuong else f"{n} không là số chính phương"

n = int(input("Nhập n: "))
print(bai_48(n))
```

## Bài 49. Số 3 chữ số `abc = a³ + b³ + c³` (số Armstrong)

```python
def bai_49():
    ket_qua = []
    for so in range(100, 1000):
        a = so // 100
        b = (so // 10) % 10
        c = so % 10
        if so == a ** 3 + b ** 3 + c ** 3:
            ket_qua.append(f"{a} {b} {c}")
    return "\n".join(ket_qua)

print(bai_49())
```

## Bài 50. Số 3 chữ số thoả `a + b + c = a·b·c`

```python
def bai_50():
    ket_qua = []
    for so in range(100, 1000):
        a = so // 100
        b = (so // 10) % 10
        c = so % 10
        if a + b + c == a * b * c:
            ket_qua.append(f"{a} {b} {c}")
    return "\n".join(ket_qua)

print(bai_50())
```
