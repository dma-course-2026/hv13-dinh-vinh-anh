## Bài 9 - Tính tiền điện

**Yêu cầu:** Tính tiền điện theo ba bậc: 60 kW đầu giá 5, 100 kW tiếp theo giá 8, phần còn lại giá 10.

```python
def bill(ho_ten, chiso1, chiso2):
    so_dien = chiso2 - chiso1
    if so_dien < 0:
        return "Chỉ số tháng này phải lớn hơn hoặc bằng chỉ số tháng trước."
    if so_dien <= 60:
        tien_dien = so_dien * 5
    elif so_dien <= 160:
        tien_dien = 60 * 5 + (so_dien - 60) * 8
    else:
        tien_dien = 60 * 5 + 100 * 8 + (so_dien - 160) * 10
    return f"Họ và tên: {ho_ten}\nTiền phải trả là   {tien_dien:.2f}"

ho_ten = input("Họ và tên: ")
chiso1, chiso2 = map(int, input("Nhập chỉ số tháng trước, chỉ số tháng này: ").split())
print(bill(ho_ten, chiso1, chiso2))
```

## Bài 14 - Hình tam giác

**Yêu cầu:** Kiểm tra tam giác và tính chu vi, diện tích bằng công thức Heron.

```python
import math

def triangle(a, b, c):
    if (a + b > c) and (a + c > b) and (b + c > a):
        perimeter = a + b + c
        semi_perimeter = perimeter / 2
        area = math.sqrt(semi_perimeter * (semi_perimeter - a) * (semi_perimeter - b) * (semi_perimeter - c))
    return perimeter, round(area, 2)

a, b, c = map(float, input("Nhập 3 số a, b, c: ").split())
print(triangle(a, b, c))
```

## Bài 17 - Tam giác vuông

**Yêu cầu:** Kiểm tra tam giác vuông bằng định lý Pythagoras.

```python
def right_triagle_check(a, b, c):
  if not (a + b > c and a + c > b and b + c > a):
    return False
  if (
      a * a == b * b + c * c
      or b * b == a * a + c * c
      or c * c == a * a + b * b
  ):
    return True
  return False
    
a = float(input("Nhập a: "))
b = float(input("Nhập b: "))
c = float(input("Nhập c: "))
print(right_triagle_check(a, b, c))
```

## Bài 20 - Phương trình bậc 2

**Yêu cầu:** Giải phương trình ax^2 + bx + c = 0 theo delta.

```python
import math

def solve_quadratic(a, b, c):
    if a == 0:
        return solve_linear(b, c)
    delta = b ** 2 - 4 * a * c
    if delta < 0:
        return "vô nghiệm"
    if delta == 0:
        x = -b / (2 * a)
        return (round(x, 2))
    x1 = (-b + math.sqrt(delta)) / (2 * a)
    x2 = (-b - math.sqrt(delta)) / (2 * a)
    return (round(x1, 2), round(x2, 2))

a = float(input("Nhập a: "))
b = float(input("Nhập b: "))
c = float(input("Nhập c: "))
print(solve_quadratic(a, b, c))
```

## Bài 28 - Thực hiện phép toán

**Yêu cầu:** Thực hiện phép toán +, -, *, / và báo lỗi cho phép toán không hợp lệ hoặc chia cho 0.

```python
def calculate(a, b, operator):
    if operator == "+":
        return a + b
    if operator == "-":
        return a - b
    if operator == "*":
        return a * b
    if operator == "/":
        if b == 0:
            raise ValueError("không thể chia cho 0")
        return round (a / b, 2)
    raise ValueError("phép toán không hợp lệ")

a = int(input("Nhập số thứ 1: "))
b = int(input("Nhập số thứ 2: "))
operator = input("Nhập phép toán (+, -, *, /): ").strip()
print(calculate(a, b, operator))
```

## Bài 33 - Điểm thuộc đường tròn

**Yêu cầu:** Phân loại điểm A nằm trong, trên hay ngoài đường tròn.

```python
import math

def circle_point_check(x0, y0, R, xa, ya):
    d = math.sqrt((xa - x0)**2 + (ya - y0)**2)
    EPSILON = 1e-9
    if abs(d - R) < EPSILON:
        return "TRÊN"
    elif d < R:
        return "TRONG"
    else:
        return "NGOÀI"

R = float(input("Nhập bán kính R: "))
x0 = float(input("Nhập x0 (tọa độ tâm): "))
y0 = float(input("Nhập y0 (tọa độ tâm): "))
xa = float(input("Nhập xa (tọa độ điểm A): "))
ya = float(input("Nhập ya (tọa độ điểm A): "))
print(circle_point_check(x0, y0, R, xa, ya))
```

## Bài 39 - Lũy thừa a^n

**Yêu cầu:** Tính a^n bằng vòng lặp.

```python
def power(a, n):
    if n == 0:
        return 1
    return a * power(a, n - 1)

a = int(input("Nhập cơ số: "))
n = int(input("Nhập số mũ: "))
print(power(a, n))
```

## Bài 40 - Tổng số chia hết cho 3 hoặc 5

**Yêu cầu:** Tính tổng các số trong [M, N] chia hết cho 3 hoặc 5.

```python
def sum_in_loop_1(m, n):
    sum = 0
    for i in range(m, n + 1):
        # Nếu i chia hết cho 3 HOẶC chia hết cho 5
        if i % 3 == 0 or i % 5 == 0:
            sum += i
    return sum

m = int(input("Nhập M: "))
n = int(input("Nhập N: "))
print(sum_in_loop_1(m, n))
```
